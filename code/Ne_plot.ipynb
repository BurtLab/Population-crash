```r
pdf("Ne.pdf", width = 6, height = 10)
par(mfrow = c(2, 1), 
    mar = c(5, 4, 3, 2) + 0.1,
    oma = c(0, 0, 0, 0))

######################## Constant model ##############################

plot(NA, xlim = c(48, 0), ylim = c(0, 33000),
     xlab = "Generations ago (0 = present)", ylab = expression(N[e]), main = "Constant")
mtext("A", side = 3, line = 1, at = par("usr")[1], 
      adj = -0.5, cex = 1.5, font = 2)

# 99% crash
lines(x = c(72, 36, 36, 0), y = c(30000, 30000, 300, 300),
      col = adjustcolor("cyan", alpha = 0.5), lwd = 2)

# 90% crash
lines(x = c(72, 36, 36, 0), y = c(30000, 30000, 3000, 3000),
      col = adjustcolor("blue", alpha = 0.5), lwd = 2)

# Annotations
abline(v = 36, col = "black", lty = 2, lwd = 1.5)
text(x = 36, y = 32000, label = "t=36", pos = 4)
text(x = 25, y = 3700, label = "90% crash", pos = 4)
text(x = 25, y = 1100, label = "99% crash", pos = 4)

######################### Seasonal model #################################

plot(NA, xlim = c(48, 0), ylim = c(0, 33000),
     xlab = "Generations ago (0 = present)", ylab = expression(N[e]), 
     main = "Seasonal")
mtext("B", side = 3, line = 1, at = par("usr")[1], 
      adj = -0.5, cex = 1.5, font = 2)

## Pre-crash (t=48 to t=36)
blue_phase1_breaks <- c(48, 42, 36)

# Horizontal segments
segments(x0 = 48, y0 = 3000, x1 = 42, y1 = 3000,
         col = adjustcolor("blue", alpha = 0.5), lwd = 2)
segments(x0 = 42, y0 = 30000, x1 = 36, y1 = 30000,
         col = adjustcolor("blue", alpha = 0.5), lwd = 2)

# Vertical segment at t=42
segments(x0 = 42, y0 = 3000, x1 = 42, y1 = 30000,
         col = adjustcolor("blue", alpha = 0.5), lwd = 2)

## Post-crash (t=36 to t=0)
blue_phase2_breaks <- seq(36, 0, by = -6)

# Horizontal segments
for (i in seq_along(blue_phase2_breaks)) {
  if (i < length(blue_phase2_breaks)) {
    x_start <- blue_phase2_breaks[i]
    x_end <- blue_phase2_breaks[i + 1]
    
    # Cycle of 300 and 3000
    if (i %% 2 == 1) {
      segments(x0 = x_start, y0 = 300, x1 = x_end, y1 = 300,
               col = adjustcolor("blue", alpha = 0.5), lwd = 2)
    } else {
      segments(x0 = x_start, y0 = 3000, x1 = x_end, y1 = 3000,
               col = adjustcolor("blue", alpha = 0.5), lwd = 2)
    }
  }
}

# Vertical segments at each break
for (x in blue_phase2_breaks[blue_phase2_breaks > 0 & blue_phase2_breaks < 36]) {
  # From 300 to 3000 at 0<t<36
  segments(x0 = x, y0 = 300, x1 = x, y1 = 3000,
           col = adjustcolor("blue", alpha = 0.5), lwd = 2)
}

# From 300 to 30000 at t=36
segments(x0 = 36, y0 = 300, x1 = 36, y1 = 30000,
         col = adjustcolor("blue", alpha = 0.5), lwd = 2)

## Annotations
abline(v = 36, col = "black", lty = 2, lwd = 1.5)
text(x = 36, y = 32000, label = "t=36", pos = 4)
text(x = 25, y = 3800, label = "90% crash", pos = 4)

dev.off()
```
